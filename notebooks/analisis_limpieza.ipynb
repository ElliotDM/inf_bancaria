<h2>Tratamiento de los datos</h2>

<h3>Eliminar datos faltantes</h3>

```python
from inf_bancaria.clean_data import remove_variables
```

```python
columns = ["CUENTA_ASIGNADA", "MESES_VENCIDOS", "SALDO_CUENTA",
           "CAPACIDAD_TC", "SEGMENTO_CLIENTE"]
```

```python
df = remove_variables(df, columns)
```

```python
df.head(10)
```

<h3>Eliminar variables innecesarias</h3>

```python
columns = ["NUM_SOLICITUD", "SUCURSAL", "PUNTO_DECISION",
            "PEOR_HISTORIA_CREDITO", "NUM_CREDITOS",]
```

```python
df = remove_variables(df, columns)
```

```python
df.head(10)
```

<h3>Corregir tipos de datos</h3>

```python
df.dtypes
```

```python
df["STATUS_SOLICITUD"] = df["STATUS_SOLICITUD"].astype(str)
df["APROBACION_TC"] = df["APROBACION_TC"].astype(str)
df["TIPO_CTE"] = df["TIPO_CTE"].astype(str)
df["COMPROBANTE_INGRESOS"] = df["COMPROBANTE_INGRESOS"].astype(str)
df["CLIENTE_CDE"] = df["CLIENTE_CDE"].astype(str)
df["NIVEL_RIESGO"] = df["NIVEL_RIESGO"].astype(str)
df["TIPO_VIVIENDA"] = df["TIPO_VIVIENDA"].astype(str)
df["ESCOLARIDAD"] = df["ESCOLARIDAD"].astype(str)
```

```python
from inf_bancaria.clean_data import remove_percent
from inf_bancaria.clean_data import normalize_percentage
```

```python
df["CAPACIDAD_PAGO_TOTAL"] = df["CAPACIDAD_PAGO_TOTAL"].astype(str)
```

```python
df['CAPACIDAD_PAGO_TOTAL'] = df['CAPACIDAD_PAGO_TOTAL'].apply(remove_percent).astype('float')
```

```python
df['CAPACIDAD_PAGO_TOTAL'] = df['CAPACIDAD_PAGO_TOTAL'].apply(normalize_percentage)
```

```python
from inf_bancaria.config import processed_data_path
```

```python
df.to_csv(processed_data_path, index=False)
```
